Consumidor de agua

```python
def calcular_tarifa_agua(consumo_mts):
    """
    Calcula el valor a pagar por el consumo de agua potable
    basado en un sistema de tarifas por tramos.
    """

    # Definimos las constantes para que el código sea más legible
    CUOTA_FIJA = 6.00
    PRECIO_TRAMO_2 = 0.45
    PRECIO_TRAMO_3 = 0.65

    LIMITE_TRAMO_1 = 18
    LIMITE_TRAMO_2 = 28

    costo_total = 0.0

    # Condicional para determinar en qué tramo se encuentra el consumo
    if consumo_mts <= 0:
        costo_total = 0.0 # No hay costo si no hay consumo
    elif consumo_mts <= LIMITE_TRAMO_1:
        # TRAMO 1: Se paga únicamente la cuota fija
        costo_total = CUOTA_FIJA
    elif consumo_mts <= LIMITE_TRAMO_2:
        # TRAMO 2: Cuota fija + costo del exceso sobre el tramo 1
        exceso_tramo_1 = consumo_mts - LIMITE_TRAMO_1
        costo_total = CUOTA_FIJA + (exceso_tramo_1 * PRECIO_TRAMO_2)
    else: # El consumo es mayor a 28
        # TRAMO 3: Se suman los costos de todos los tramos

        # Costo del primer tramo (completo)
        costo_tramo_1 = CUOTA_FIJA

        # Costo del segundo tramo (completo: 10 metros del 19 al 28)
        metros_tramo_2 = LIMITE_TRAMO_2 - LIMITE_TRAMO_1
        costo_tramo_2 = metros_tramo_2 * PRECIO_TRAMO_2

        # Costo del exceso sobre el tramo 2
        exceso_tramo_2 = consumo_mts - LIMITE_TRAMO_2
        costo_tramo_3 = exceso_tramo_2 * PRECIO_TRAMO_3

        costo_total = costo_tramo_1 + costo_tramo_2 + costo_tramo_3

    return costo_total

# --- Bloque principal para ejecutar el programa ---
if __name__ == "__main__":
    try:
        # Pedimos al usuario que ingrese los datos
        metros_consumidos = float(input("Ingrese el número de metros cúbicos consumidos: "))

        # Llamamos a la función para hacer el cálculo
        valor_a_pagar = calcular_tarifa_agua(metros_consumidos)

        # Mostramos el resultado formateado a dos decimales, como es usual en dinero
        print(f"El consumo fue de {metros_consumidos} m³.")
        print(f"El valor total a pagar es: ${valor_a_pagar:.2f}")

    except ValueError:
        # Manejo de error si el usuario no ingresa un número
        print("Error: Por favor, ingrese un valor numérico válido.")
```

```python
# --- Constantes de Conversión ---
# Unidades de Empaque/Inventario
UNIDADES_POR_CAJA = 25.0
LIBRAS_POR_SACO = 50.0
UNIDADES_POR_FARDO = 5.0

# --- NUEVAS CONSTANTES ---
# Longitud
PIES_POR_METRO = 3.28084

# Almacenamiento Digital
MEGABYTES_POR_GIGABYTE = 1024.0

# La conversión de temperatura es una fórmula, no una constante, así que se manejará directamente.

def mostrar_menu():
    """Imprime el menú de opciones ampliado para el usuario."""
    print("\n--- Conversor Universal ---")
    print("--- Empaque/Inventario ---")
    print("1. Cajas a Unidades")
    print("2. Unidades a Cajas")
    print("3. Sacos a Libras")
    print("4. Libras a Sacos")
    print("5. Fardos a Unidades")
    print("6. Unidades a Fardos")
    print("\n--- Longitud ---")
    print("7. Metros a Pies")
    print("8. Pies a Metros")
    print("\n--- Temperatura ---")
    print("9. Celsius a Fahrenheit")
    print("10. Fahrenheit a Celsius")
    print("\n--- Almacenamiento Digital ---")
    print("11. Gigabytes (GB) a Megabytes (MB)")
    print("12. Megabytes (MB) a Gigabytes (GB)")
    print("\n13. Salir")

# --- Bucle principal del programa ---
while True:
    mostrar_menu()
    opcion = input("Seleccione una opción: ")

    # Usamos un bloque try-except general para el input numérico
    try:
        # --- Conversiones de Empaque ---
        if opcion == '1':
            cajas = float(input("Ingrese la cantidad de cajas: "))
            unidades = cajas * UNIDADES_POR_CAJA
            print(f"-> {cajas} cajas son {unidades} unidades.")

        elif opcion == '2':
            unidades = float(input("Ingrese la cantidad de unidades: "))
            cajas = unidades / UNIDADES_POR_CAJA
            print(f"-> {unidades} unidades son {cajas:.2f} cajas.")

        elif opcion == '3':
            sacos = float(input("Ingrese la cantidad de sacos: "))
            libras = sacos * LIBRAS_POR_SACO
            print(f"-> {sacos} sacos son {libras} libras.")

        elif opcion == '4':
            libras = float(input("Ingrese la cantidad de libras: "))
            sacos = libras / LIBRAS_POR_SACO
            print(f"-> {libras} libras son {sacos:.2f} sacos.")

        elif opcion == '5':
            fardos = float(input("Ingrese la cantidad de fardos: "))
            unidades = fardos * UNIDADES_POR_FARDO
            print(f"-> {fardos} fardos son {unidades} unidades.")

        elif opcion == '6':
            unidades = float(input("Ingrese la cantidad de unidades: "))
            fardos = unidades / UNIDADES_POR_FARDO
            print(f"-> {unidades} unidades son {fardos:.2f} fardos.")

        # --- NUEVAS CONVERSIONES ---
        # --- Longitud ---
        elif opcion == '7':
            metros = float(input("Ingrese la cantidad de metros: "))
            pies = metros * PIES_POR_METRO
            print(f"-> {metros} metros son {pies:.2f} pies.")

        elif opcion == '8':
            pies = float(input("Ingrese la cantidad de pies: "))
            metros = pies / PIES_POR_METRO
            print(f"-> {pies} pies son {metros:.2f} metros.")

        # --- Temperatura (usando fórmulas) ---
        elif opcion == '9':
            celsius = float(input("Ingrese los grados Celsius: "))
            fahrenheit = (celsius * 9/5) + 32
            print(f"-> {celsius}°C son {fahrenheit:.2f}°F.")

        elif opcion == '10':
            fahrenheit = float(input("Ingrese los grados Fahrenheit: "))
            celsius = (fahrenheit - 32) * 5/9
            print(f"-> {fahrenheit}°F son {celsius:.2f}°C.")

        # --- Almacenamiento Digital ---
        elif opcion == '11':
            gb = float(input("Ingrese la cantidad de Gigabytes (GB): "))
            mb = gb * MEGABYTES_POR_GIGABYTE
            print(f"-> {gb} GB son {mb} MB.")

        elif opcion == '12':
            mb = float(input("Ingrese la cantidad de Megabytes (MB): "))
            gb = mb / MEGABYTES_POR_GIGABYTE
            print(f"-> {mb} MB son {gb:.2f} GB.")

        # --- Opción de Salida ---
        elif opcion == '13':
            print("¡Gracias por usar el conversor! Adiós.")
            break # Rompe el bucle while y termina el programa

        else:
            print("Opción no válida. Por favor, intente de nuevo.")

    except ValueError:
        print("Error: Entrada inválida. Por favor, ingrese solo números.")
```

```python
def calcular_deducciones(sueldo):

    isss = min(sueldo * 0.03,30.00)
    afp = sueldo * 0.0725

    sueldo_gravado = sueldo - (isss + afp)

    if sueldo_gravado <= 550.00:
        isr = 0
    elif sueldo_gravado <= 895.24:
        isr = (sueldo_gravado - 550.00) * 0.10 + 17.67
    elif sueldo_gravado <= 2038.10:
        isr = (sueldo_gravado - 895.24) * 0.20 + 60.00
    else:
        isr = (sueldo_gravado - 2038.10) * 0.30 + 288.57

    if sueldo <= 0:
        print("Error: Digite un Numero Positivo")
        return (0, 0, 0, 0, 0)

    total_deducciones = isss + afp + isr
    sueldo_neto = sueldo - total_deducciones

    return (isss, afp, isr, total_deducciones, sueldo_neto)


sueldo = float(input("Ingrese su sueldo: "))
resultado = calcular_deducciones(sueldo)

print("\n--- RESULTADOS ---")
print(f"ISSS (3%): ${resultado[0]:.2f}")
print(f"AFP (7.25%): ${resultado[1]:.2f}")
print(f"ISR: ${resultado[2]:.2f}")
print(f"Total deducciones: ${resultado[3]:.2f}")
print(f"Sueldo neto: ${resultado[4]:.2f}")
```

```python
from math import sqrt
from collections import Counter
class Estadistica:
   def media(self, serie):
       suma = 0
       for num in serie:
           suma += int(num)
       return suma / len(serie)

   def varianza(self, serie):
       media = self.media(serie)
       suma = 0
       for num in serie:
           suma += (int(num) - media) ** 2
       return suma / len(serie)
   def frecuencia(self, serie):
       return Counter(serie)
```

```python
# Entrada de la serie de números
serie = input("Serie (ejemplo: 1,5,2,3): ").split(",")
objEstadistica = Estadistica()

# Cálculo de media, varianza y frecuencia
print("La media es: ", objEstadistica.media(serie))
print("La varianza es: ", round(objEstadistica.varianza(serie), 2))
print("La frecuencia es: ", objEstadistica.frecuencia(serie))
```

```python
def calcular_impuesto_actividades_economicas(monto_actividad):
    """
    Calcula el IMPUESTO A LAS ACTIVIDADES ECONÓMICAS de acuerdo con la tabla dada.
    """
    tabla_impuestos = [
        (0.01, 500, 1.5, 0),
        (500.01, 1000, 1.5, 3),
        (1000.01, 2000, 3, 3),
        (2000.01, 3000, 6, 3),
        (3000.01, 6000, 9, 2),
        (8000.01, 18000, 15, 2),
        (18000.01, 30000, 39, 2),
        (30000.01, 60000, 63, 1),
        (60000.01, 100000, 93, 0.8),
        (100000.01, 200000, 125, 0.7),
        (200000.01, 300000, 195, 0.6),
        (300000.01, 400000, 255, 0.45),
        (400000.01, 500000, 300, 0.4),
        (500000.01, 1000000, 340, 0.30),
        (1000000.01, 99999999, 490, 0.18)
    ]

    for desde, hasta, precio, adicional in tabla_impuestos:
        if desde <= monto_actividad <= hasta:
            if adicional == 0:
                return precio
            else:
                impuesto = ((monto_actividad - desde) / 1000 * adicional) + precio
                return round(impuesto, 2)
    return "Monto fuera de los rangos definidos."

def convertir_area(cantidad, origen, destino):
    """
    Convierte unidades de área basadas en las medidas de El Salvador.
    """
    unidades = {
        "pie cuadrado": 0.092903,
        "vara cuadrada": 0.6987,
        "yarda cuadrada": 0.836127,
        "metro cuadrado": 1.0,
        "tarea": 437.5, # Valor corregido: 1/16 de una manzana.
        "manzana": 7000.0, # 7,000 m2
        "hectárea": 10000.0
    }

    origen_lower = origen.lower()
    destino_lower = destino.lower()

    if origen_lower not in unidades or destino_lower not in unidades:
        return None

    cantidad_en_m2 = cantidad * unidades[origen_lower]
    resultado = cantidad_en_m2 / unidades[destino_lower]
    return resultado

def simular_cajero(monto_a_retirar):
    """
    Simula la dispensación de billetes y monedas.
    """
    try:
        cantidad = float(monto_a_retirar)
    except ValueError:
        print("Entrada inválida. Por favor, ingrese un número.")
        return

    if cantidad <= 0:
        print("El monto a retirar debe ser un número positivo.")
        return

    print(f"\nDesglose para ${cantidad:.2f}:")
    denominaciones_billetes = [100, 50, 20, 10, 5, 1]
    denominaciones_monedas = [0.25, 0.10, 0.05, 0.01]
    cantidad_restante = cantidad

    for billete in denominaciones_billetes:
        if cantidad_restante >= billete:
            cantidad_billetes = int(cantidad_restante // billete)
            print(f"{cantidad_billetes} billete(s) de a ${billete}")
            cantidad_restante %= billete

    cantidad_restante_centavos = round(cantidad_restante * 100)
    for moneda_centavos in [int(m * 100) for m in denominaciones_monedas]:
        if cantidad_restante_centavos >= moneda_centavos:
            cantidad_monedas = int(cantidad_restante_centavos // moneda_centavos)
            valor_moneda_dolar = moneda_centavos / 100
            print(f"{cantidad_monedas} moneda(s) de a ${valor_moneda_dolar:.2f} ctvs.")
            cantidad_restante_centavos %= moneda_centavos

def main():
    while True:
        print("\n--- Parcial I INGENIERIA EDU-ANA ---")
        print("1. Cálculo de Impuesto ")
        print("2. Conversor de Área ")
        print("3. Cajero Automático")
        print("4. Exit")
        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            try:
                monto_input = float(input("Ingrese el monto de la actividad económica: $"))
                impuesto = calcular_impuesto_actividades_economicas(monto_input)
                if isinstance(impuesto, str):
                    print(f"Error: {impuesto}")
                else:
                    print(f"El valor a pagar es: ${impuesto:.2f}")
            except ValueError:
                print("Entrada inválida. Por favor, ingrese un número.")
        elif opcion == "2":
            print("Unidades disponibles: Pie Cuadrado, Vara Cuadrada, Yarda Cuadrada, Metro Cuadrado, Tarea, Manzana, Hectárea.")
            try:
                cantidad_area = float(input("Ingrese la cantidad a convertir: "))
                unidad_origen = input("Ingrese la unidad de origen: ")
                unidad_destino = input("Ingrese la unidad de destino: ")
                resultado_area = convertir_area(cantidad_area, unidad_origen, unidad_destino)
                if resultado_area is None:
                    print("Una o ambas unidades de área no son válidas. Por favor, verifique.")
                else:
                    print(f"{cantidad_area} {unidad_origen.capitalize()} = {resultado_area:.4f} {unidad_destino.capitalize()}")
            except ValueError:
                print("Entrada inválida. Por favor, ingrese un número para la cantidad.")
        elif opcion == "3":
            try:
                monto_retirar = float(input("Ingrese la cantidad a retirar: $"))
                simular_cajero(monto_retirar)
            except ValueError:
                print("Entrada inválida. Por favor, ingrese un número.")
        elif opcion == "4":
            print("Gracias por usar el programa.")
            break
        else:
            print("Opción inválida. Intente nuevamente.")

if __name__ == "__main__":
    main()
```
